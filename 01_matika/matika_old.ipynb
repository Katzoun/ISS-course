```python
# -------------- Math for ISS --------------------------
import numpy as np
import matplotlib.pyplot as plt

# >>>>>>>>>>>>>>>>> #basicops <<<<<<<<<<<<<<<<<<<<<<<<<
# --- addition of signals ---
t = np.arange(-5,5,0.01)
x1 = np.zeros(t.shape)
ii = np.argwhere((t>-1) & (t<0.5)) 
x1[ii] = 0.5
x2 = np.zeros(t.shape)
ii = np.argwhere((t>-0.5) & (t<2)) 
x2[ii] = 1
y = x1 + x2
_, ax = plt.subplots(3,1, figsize=(10,5))
ax[0].plot(t,x1)
ax[0].set_ylabel('$x_1(t)$')
ax[0].set_ylim(-0.1,2)
ax[0].grid()
ax[1].plot(t,x2,color='g')
ax[1].set_ylabel('$x_2(t)$')
ax[1].set_ylim(-0.1,2)
ax[1].grid()
ax[2].plot(t,y,color='r',linewidth=3)
ax[2].set_ylabel('$y(t)$')
ax[2].set_xlabel('$t$')
ax[2].set_ylim(-0.1,2)
ax[2].grid()
#plt.savefig('FIG/soucet.pdf')
```

```python
# --- difference of signals ---
t = np.arange(-5,5,0.01)
x1 = np.zeros(t.shape)
ii = np.argwhere((t>-1) & (t<0.5)) 
x1[ii] = 0.5
x2 = np.zeros(t.shape)
ii = np.argwhere((t>-0.5) & (t<2)) 
x2[ii] = 1
y = x1 - x2
_, ax = plt.subplots(3,1, figsize=(10,5))
ax[0].plot(t,x1)
ax[0].set_ylabel('$x_1(t)$')
ax[0].set_ylim(-1.1,1.1)
ax[0].grid()
ax[1].plot(t,x2,color='g')
ax[1].set_ylabel('$x_2(t)$')
ax[1].set_ylim(-1.1,1.1)
ax[1].grid()
ax[2].plot(t,y,color='r',linewidth=3)
ax[2].set_ylabel('$y(t)$')
ax[2].set_xlabel('$t$')
ax[2].set_ylim(-1.1,1.1)
ax[2].grid()
#plt.savefig('FIG/rozdil.pdf')
```

```python
# --- product signalu ---
t = np.arange(-5,5,0.01)
x1 = np.zeros(t.shape)
ii = np.argwhere((t>-1) & (t<0.5)) 
x1[ii] = 0.5
x2 = np.zeros(t.shape)
ii = np.argwhere((t>-0.5) & (t<2)) 
x2[ii] = 1
y = x1 * x2
_, ax = plt.subplots(3,1, figsize=(10,5))
ax[0].plot(t,x1)
ax[0].set_ylabel('$x_1(t)$')
ax[0].set_ylim(-0.1,1.1)
ax[0].grid()
ax[1].plot(t,x2,color='g')
ax[1].set_ylabel('$x_2(t)$')
ax[1].set_ylim(-0.1,1.1)
ax[1].grid()
ax[2].plot(t,y,color='r',linewidth=3)
ax[2].set_ylabel('$y(t)$')
ax[2].set_xlabel('$t$')
ax[2].set_ylim(-0.1,1.1)
ax[2].grid()
#plt.savefig('FIG/soucin.pdf')
```

```python
# >>>>>>>>>>>>>>>>> #linear <<<<<<<<<<<<<<<<<<<<<<<<<
t = np.arange(-5,5,0.01)
x1 = 3 * t; 
x2 = 2*t + 1; 
x3 = -t; 
x4 = -2*t -2; 
plt.figure(figsize=(10,5))
plt.plot(t,x1,label="$x_1(t) = 3t$")
plt.plot(t,x2,label="$x_2(t) = 2t+1$") 
plt.plot(t,x3,label="$x_3(t) = -t$") 
plt.plot(t,x4,label="$x_4(t) = -2t-2$") 
plt.gca().set_xlabel('$t$')
plt.gca().set_ylabel('$x(t)$')
plt.gca().grid()
plt.legend()
#plt.savefig('FIG/lin.pdf')
```
